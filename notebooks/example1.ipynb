```python
# Simple example for generating predictions with MHCflurry.
import mhcflurry
```

```python
# Load a predictor
predictor = mhcflurry.Class1PresentationPredictor.load()
predictor
```

```python
# Predict peptides with their N/C source-protein flanks.
# These windows are from docs/example.fasta.
peptides = ["TPVCPNGPG", "RLLEGMEMI"]
alleles = {"sample1": ["HLA-A*02:01", "HLA-A*03:01"]}
results1 = predictor.predict(
    peptides, alleles,
    n_flanks=["MSSSS", "MVENK"],
    c_flanks=["NCQV", "FGQVI"],
)
results1
```

```python
# Compare the same peptides and MHC alleles without flanking context.
results_no_flanks = predictor.predict(peptides, alleles)
results_no_flanks
```

```python
# Predict across protein sequences
# Try help(predictor.predict_sequences) for other options.
results2 = predictor.predict_sequences(
    sequences={
        'protein1': "MDSKGSSQKGSRLLLLLVVSNLL",
        'protein2': "SSLPTPEDKEQAQQTHH",
    },
    alleles={
        "sample1": ["A0201", "A0301", "B0702"],
        "sample2": ["A0101", "C0202"],
    },
    result="filtered",
    comparison_quantity="affinity",
    filter_value=500)
results2
```
